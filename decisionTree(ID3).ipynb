```python
# ===============================
# DECISION TREE - ID3 (CLASSIFICATION)
# ===============================

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, accuracy_score

print("===== DECISION TREE (ID3) =====")

# =====================================================
# DATASET 1: PLAY TENNIS
# =====================================================

data1 = {
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast",
                "Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],
    "Temperature": ["Hot","Hot","Hot","Mild","Cool","Cool","Cool",
                    "Mild","Cool","Mild","Mild","Mild","Hot","Mild"],
    "Humidity": ["High","High","High","High","Normal","Normal","Normal",
                 "High","Normal","Normal","Normal","High","Normal","High"],
    "Wind": ["Weak","Strong","Weak","Weak","Weak","Strong","Strong",
             "Weak","Weak","Weak","Strong","Strong","Weak","Strong"],
    "PlayTennis": ["No","No","Yes","Yes","Yes","No","Yes",
                   "No","Yes","Yes","Yes","Yes","Yes","No"]
}

df1 = pd.DataFrame(data1)

# Encode categorical data
le = LabelEncoder()
for col in df1.columns:
    df1[col] = le.fit_transform(df1[col])

X1 = df1.drop("PlayTennis", axis=1)
y1 = df1["PlayTennis"]

# Train ID3 model
dt1 = DecisionTreeClassifier(criterion="entropy")
dt1.fit(X1, y1)

# Predictions
y_pred1 = dt1.predict(X1)

print("\nPlay Tennis Confusion Matrix:")
print(confusion_matrix(y1, y_pred1))
print("Accuracy:", accuracy_score(y1, y_pred1))

# Plot decision tree
plt.figure(figsize=(18,10))
plot_tree(dt1,
          feature_names=X1.columns,
          class_names=["No","Yes"],
          filled=True)
plt.title("Decision Tree (ID3) - Play Tennis Dataset")
plt.show()

# =====================================================
# DATASET 2: STUDENT PERFORMANCE
# =====================================================

data2 = {
    "Attendance": ["High","High","Medium","Low","Low",
                   "Medium","High","Medium","Low","High"],
    "StudyHours": ["High","Medium","High","Medium","Low",
                   "Low","Medium","Medium","High","High"],
    "Sleep": ["Good","Good","Poor","Poor","Good",
              "Good","Poor","Good","Poor","Good"],
    "Result": ["Pass","Pass","Pass","Fail","Fail",
               "Fail","Pass","Pass","Fail","Pass"]
}

df2 = pd.DataFrame(data2)

# Encode categorical data
for col in df2.columns:
    df2[col] = le.fit_transform(df2[col])

X2 = df2.drop("Result", axis=1)
y2 = df2["Result"]

# Train ID3 model
dt2 = DecisionTreeClassifier(criterion="entropy")
dt2.fit(X2, y2)

# Predictions
y_pred2 = dt2.predict(X2)

print("\nStudent Performance Confusion Matrix:")
print(confusion_matrix(y2, y_pred2))
print("Accuracy:", accuracy_score(y2, y_pred2))

# Plot decision tree
plt.figure(figsize=(16,8))
plot_tree(dt2,
          feature_names=X2.columns,
          class_names=["Fail","Pass"],
          filled=True)
plt.title("Decision Tree (ID3) - Student Performance Dataset")
plt.show()
```
