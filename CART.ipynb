```python
# ===============================
# CART REGRESSION TREE
# ===============================

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error

print("===== CART REGRESSION TREE =====")

# =====================================================
# DATASET 1: HOUSE PRICE PREDICTION
# =====================================================

# Features: Size (sqft), Bedrooms
# Target: Price (Lakhs)

X1 = [[800,1],
      [900,2],
      [1000,2],
      [1200,3],
      [1500,3],
      [1800,4]]

y1 = [40,45,50,60,75,90]

# Train CART model
cart1 = DecisionTreeRegressor()
cart1.fit(X1, y1)

# Prediction
y_pred1 = cart1.predict(X1)

# Error matrix
error_df1 = pd.DataFrame({
    "Actual Price": y1,
    "Predicted Price": y_pred1,
    "Error": [a - p for a, p in zip(y1, y_pred1)]
})
error_df1["Error^2"] = error_df1["Error"]**2

print("\nHouse Price Error Matrix:")
print(error_df1)
print("MSE:", mean_squared_error(y1, y_pred1))

# Plot tree
plt.figure(figsize=(16,8))
plot_tree(cart1,
          feature_names=["Size", "Bedrooms"],
          filled=True)
plt.title("CART Regression Tree - House Price Prediction")
plt.show()

# =====================================================
# DATASET 2: SALARY PREDICTION
# =====================================================

# Feature: Experience (Years)
# Target: Salary (LPA)

X2 = [[1],[2],[3],[4],[5],[6]]
y2 = [3,4,5,7,9,11]

# Train CART model
cart2 = DecisionTreeRegressor()
cart2.fit(X2, y2)

# Prediction
y_pred2 = cart2.predict(X2)

# Error matrix
error_df2 = pd.DataFrame({
    "Actual Salary": y2,
    "Predicted Salary": y_pred2,
    "Error": [a - p for a, p in zip(y2, y_pred2)]
})
error_df2["Error^2"] = error_df2["Error"]**2

print("\nSalary Prediction Error Matrix:")
print(error_df2)
print("MSE:", mean_squared_error(y2, y_pred2))

# Plot tree
plt.figure(figsize=(14,8))
plot_tree(cart2,
          feature_names=["Experience"],
          filled=True)
plt.title("CART Regression Tree - Salary Prediction")
plt.show()
```
